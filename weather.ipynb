```python
import os, requests
from dotenv import load_dotenv

load_dotenv()

API_KEY = os.getenv("API_KEY")
API_URL = "https://api.weatherapi.com/v1/current.json"

import pandas as pd
```

```python
def get_weather(city, url=API_URL, key=API_KEY):
    data = {}
    parameters = {'key':key, 'q':city}
    response = requests.get(url, params=parameters)
    
    # Check if the request was successful
    if response.status_code == 200:
        raw = response.json()
        data['city'] = raw['location']['name']
        data['temperature_f'] = raw['current']['temp_f']
        data['temperature_c'] = raw['current']['temp_c']
        data['feels_like_f'] = raw['current']['feelslike_f']
        data['feels_like_c'] = raw['current']['feelslike_c']
        data['humidity'] = raw['current']['humidity']
        data['wind'] = raw['current']['wind_mph']
        data['condition'] = raw['current']['condition']['text']
        data['last_updated'] = raw['current']['last_updated']
    else:
        print(f"Error fetching {city}.\n Response: {response.status_code}")
    return data
```

```python
cities = ["Matanzas", "Louisville", "Miami", "Montevideo", "Denver"]
bulk_data = []
for city in cities:
    bulk_data.append(get_weather(city))
```

```python
weather_df = pd.DataFrame(bulk_data)
```
